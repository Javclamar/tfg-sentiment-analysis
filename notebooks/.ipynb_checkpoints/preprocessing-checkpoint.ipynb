```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import csv
```

```python
input_json_reviews = '../data/json/yelp_academic_dataset_review.json'
output_csv_reviews = '../data/csv/yelp_academic_dataset_review.csv'
chunk_size = 100000

def convertir_json_a_csv(input_file: str, output_file: str, chunk_size: int = 100000):
    """
    Lee un archivo JSON por bloques y lo convierte a CSV.
    
    Args:
        input_file (str): Ruta al archivo JSON original (Yelp).
        output_file (str): Ruta donde se guardará el CSV resultante.
        chunk_size (int): Número de filas a procesar por iteración.
    """
    
    reader = pd.read_json(input_file, lines=True, chunksize=chunk_size)
    
    first_chunk = True
    
    for chunk in reader:
        
        if first_chunk:
            #Eliminamos las filas que tengan valores nulos en las columnas que se muestran
            chunk.dropna(subset=['text', 'stars', 'useful', 'funny', 'cool', 'date')
            # Modo 'w' (write) para crear el archivo y escribir cabeceras
            chunk.to_csv(output_file, mode='w', index=False, encoding='utf-8')
            first_chunk = False
        else:
            # Modo 'a' (append) para añadir al final sin cabeceras
            chunk.to_csv(output_file, mode='a', index=False, header=False, encoding='utf-8')

convertir_json_a_csv(input_json_reviews, output_csv_reviews, chunk_size)
```

```python
input_csv_reviews = '../data/csv/yelp_academic_dataset_review.csv'
sample_csv_reviews = '../data/csv/yelp_academic_dataset_review_sample.csv'
size = 100000

def obtener_muestra_aleatoria(input_csv_file: str, output_csv_file: str, size: int):
    """
    Obtiene una muestra aleatoria de un dataset almacenado en un csv para realizar estudios y pruebas sobre esta.

    Args:
        inptut_csv_file (str): Ruta al archivo csv.
        output_csv_file (str): Ruta donde se guardará la muestra resultante.
        size (int): Número máximo de filas de la muestra.
    """

    # Se define la fracción de los datos que van a componer la muestra
    sample_list=[]
    fraction = 0.01

    for chunk in pd.read_csv(input_csv_file, chunksize=100000):
        sample_list.append(chunk.sample(frac=fraction, random_state=42))

    df_sample = pd.concat(sample_list, ignore_index=True)

    # Comprobar si la muestra es más grande de lo indicado en size
    if len(df_sample) > size:
        df_sample = df_sample.sample(n=size, random_state=42)

    # Crea el csv en la ruta indicada
    df_sample.to_csv(output_csv_file, mode='w', index=False, encoding='utf-8')
    print(f"Muestra creada con {len(df_sample)} filas y guardada en {output_csv_file}")
    display(df_sample.head())
    
obtener_muestra_aleatoria(csv_reviews, sample_csv_reviews, size)
```

```python
sample_csv_reviews = '../data/csv/yelp_academic_dataset_review_sample.csv'

def analisis_exploratorio(csv_sample: str):
    """
    Realiza un EDA sobre la muestra que se indica, realizando un sanity check, la distribución de stars y buscando patrones

    Args:
        csv_sample (str): Ruta del fichero csv que contiene la muestra
    """
    
    df = pd.read_csv(csv_sample)

    # Realizamos un sanity check a la muestra buscando nulos y información general
    print(f"--- Información General ---")
    print(f"\n{df.info()}")
    print(f"\nDuplicados: {df.duplicated().sum()}")
    display(df.describe())

    # Mostramos la distribución del Target (stars)
    plt.figure(figsize=(8, 5))
    sns.countplot(x='stars', hue='stars', data=df, palette='viridis', legend=False)    
    plt.title('Distribución de Estrellas en la Muestra')
    plt.xlabel('Estrellas')
    plt.ylabel('Cantidad de Reseñas')
    plt.show()
  
    # Ver los porcentajes exactos
    print(df['stars'].value_counts(normalize=True) * 100)


analisis_exploratorio(sample_csv_reviews)
```
